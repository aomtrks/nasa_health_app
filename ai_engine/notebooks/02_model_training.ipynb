```python
import numpy as np, json, tensorflow as tf

def run(path, x):
    it = tf.lite.Interpreter(model_path=path)
    it.allocate_tensors()
    i = it.get_input_details()[0]; o = it.get_output_details()[0]
    it.set_tensor(i['index'], x.astype(np.float32))
    it.invoke()
    return it.get_tensor(o['index'])
```

```python
sc = json.load(open('cardiovascular_scaler.json'))
raw = np.array([68, 2, 181, 83, 114, 72, 1, 1, 0, 0, 1, 76/1.78**2])
x = ((raw - np.array(sc['mean'])) / np.array(sc['scale']))[None, :]
print(run('cardiovascular.tflite', x))
```
